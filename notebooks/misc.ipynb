```python
import re

test_df = pd.read_csv("data/test_rewrite_002.csv")

l = []
for idx, query in enumerate(test_df['query'].tolist()):
    if idx >= 4:
        break
    l.append(query)

tokens0 = set(re.split(r'\s+', l[0].lower()))
tokens1 = set(re.split(r'\s+', l[1].lower()))
print(len(tokens0), len(tokens1), len(tokens0 & tokens1))

tokens1 = set(re.split(r'\s+', l[1].lower()))
tokens2 = set(re.split(r'\s+', l[2].lower()))
print(len(tokens1), len(tokens2), len(tokens1 & tokens2))

tokens2 = set(re.split(r'\s+', l[2].lower()))
tokens3 = set(re.split(r'\s+', l[3].lower()))
print(len(tokens2), len(tokens3), len(tokens2 & tokens3))
    
```
